```python
import pandas as pd
```

```python
import os
```

```python
DATA_PATH = "../data/raw"
```

```python
customers = pd.read_csv(f"{DATA_PATH}/blinkit_customers.csv")
products =pd.read_csv(f"{DATA_PATH}/blinkit_products.csv")
orders = pd.read_csv(f"{DATA_PATH}/blinkit_orders.csv")
order_items = pd.read_csv(f"{DATA_PATH}/blinkit_order_items.csv")
inventory = pd.read_csv(f"{DATA_PATH}/blinkit_inventory.csv")
delivery = pd.read_csv(f"{DATA_PATH}/blinkit_delivery_performance.csv")
marketing = pd.read_csv(f"{DATA_PATH}/blinkit_marketing_performance.csv")
feedback = pd.read_csv(f"{DATA_PATH}/blinkit_customer_feedback.csv")
```

```python
def inspect_dataset(df, name):
    print("="*60)
    print(f"Dataset : {name}")
    print("="*60)

    print("\nShape")
    print(df.shape)

    print("\nColumns")
    print(df.columns.tolist())

    print("\nInfo")
    print(df.info())

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())

    print("\nFirst Five Rows")
    print(df.head())

    print("\n")
```

```python
inspect_dataset(customers, "Customers")
inspect_dataset(products, "Products")
inspect_dataset(orders, "Orders")
inspect_dataset(order_items, "Order Items")
inspect_dataset(inventory, "Inventory")
inspect_dataset(delivery, "Delivery")
inspect_dataset(marketing, "Marketing")
inspect_dataset(feedback, "Feedback")
```
