```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
RAW_PATH = Path("../data/raw")
CLEANED_PATH = Path("../data/cleaned")
CLEANED_PATH.mkdir(exist_ok=True)
```

```python
customers = pd.read_csv(RAW_PATH / "blinkit_customers.csv")
products = pd.read_csv(RAW_PATH / "blinkit_products.csv")
orders = pd.read_csv(RAW_PATH / "blinkit_orders.csv")
order_items = pd.read_csv(RAW_PATH / "blinkit_order_items.csv")
inventory = pd.read_csv(RAW_PATH / "blinkit_inventory.csv")
delivery = pd.read_csv(RAW_PATH / "blinkit_delivery_performance.csv")
marketing = pd.read_csv(RAW_PATH / "blinkit_marketing_performance.csv")
feedback = pd.read_csv(RAW_PATH / "blinkit_customer_feedback.csv")
```

```python
customers["registration_date"] = pd.to_datetime(customers["registration_date"])
orders["order_date"] = pd.to_datetime(orders["order_date"])
orders["promised_delivery_time"] = pd.to_datetime(orders["promised_delivery_time"])
orders["actual_delivery_time"] = pd.to_datetime(orders["actual_delivery_time"])
inventory["date"] = pd.to_datetime(inventory["date"], dayfirst=True)

delivery["promised_time"] =pd.to_datetime(delivery["promised_time"])
delivery["actual_time"] = pd.to_datetime(delivery["actual_time"])

marketing["date"] = pd.to_datetime(marketing["date"])

feedback["feedback__date"] =pd.to_datetime(feedback["feedback_date"])
```

```python
customers.info()
orders.info()
inventory.info()
delivery.info()
marketing.info()
feedback.info()
```

```python
def check_primary_key(df, column):
    unique = df[column].is_unique
    print(f"{column}: {'PASS' if unique else 'FAIL'}")
```

```python
check_primary_key(customers,"customer_id")
check_primary_key(products,"product_id")
check_primary_key(orders,"order_id")
check_primary_key(marketing,"campaign_id")
check_primary_key(feedback,"feedback_id")
```

```python
invalid_customers = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print(len(invalid_customers))
```

```python
invalid_products = order_items[
    ~order_items["product_id"].isin(products["product_id"])
]

print(len(invalid_products))
```

```python
invalid_orders = order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]

print(len(invalid_orders))
```

```python
customers.to_csv(CLEANED_PATH / "customers_cleaned.csv", index=False)
products.to_csv(CLEANED_PATH / "products_cleaned.csv", index=False)
orders.to_csv(CLEANED_PATH / "orders_cleaned.csv", index=False)
order_items.to_csv(CLEANED_PATH / "order_items_cleaned.csv", index=False)
inventory.to_csv(CLEANED_PATH / "inventory_cleaned.csv", index=False)
delivery.to_csv(CLEANED_PATH / "delivery_cleaned.csv", index=False)
marketing.to_csv(CLEANED_PATH / "marketing_cleaned.csv", index=False)
feedback.to_csv(CLEANED_PATH / "feedback_cleaned.csv", index=False)
```
